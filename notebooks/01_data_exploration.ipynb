```python
# Import and Load dataset
import pandas as pd
from pathlib import Path

# Define the dataset
DATA_PATH = Path("../data/raw/Flight_Price_Dataset_of_Bangladesh.csv")
DATA_PATH
```

```python
# Load the raw dataset
df = pd.read_csv(DATA_PATH)

df.head()
```

```python
df.info()
```

```python
# Data validation checks

required_columns = [
    "Airline",
    "Source",
    "Destination",
    "Base Fare (BDT)",
    "Tax & Surcharge (BDT)",
    "Total Fare (BDT)",
]

# 1. Required columns
missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

# 2. Empty/null values
null_counts = df[required_columns].isna().sum()

# 3. Negative fares
negative_fares = (
    (df["Base Fare (BDT)"] < 0)
    | (df["Tax & Surcharge (BDT)"] < 0)
    | (df["Total Fare (BDT)"] < 0)
).sum()

# 4. Fare calculation consistency
fare_mismatch = (
    (df["Base Fare (BDT)"] + df["Tax & Surcharge (BDT)"])
    .round(2)
    != df["Total Fare (BDT)"].round(2)
).sum()

# 5. Empty categorical values
empty_categories = (
    df[["Airline", "Source", "Destination"]]
    .astype(str)
    .apply(lambda col: col.str.strip().eq("").sum())
)

print("Missing required columns:", missing_columns)
print("\nNull values:")
print(null_counts)
print("\nNegative fare records:", negative_fares)
print("Fare calculation mismatches:", fare_mismatch)
print("\nEmpty categorical values:")
print(empty_categories)
```
